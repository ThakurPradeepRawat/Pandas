## series

 It is a one dimensional  labelled array capable of strong / holding homogeneous data of any type (int,str,float, _ _ _ _ _ , python objects etc)

Pandas series is nothing but a column value in excel-sheet

Pandas series values are mutable

### creating a series object 

Syntax :- varname = pandas.Series(object,index,dtype)

```python
# creating series object by int type
a = 100
var = pd.Series(a)
print(var,type(var))
```

```python
# creating series object by float type
a = 100.89
var = pd.Series(a)
print(var,type(var))
```

```python
# creating series object by bool type
a = True
var = pd.Series(a)
print(var,type(var))
```

```python
# creating series object by complex type
a = 2+3j
var = pd.Series(a)
print(var,type(var))
```

```python
# creating series object by str type
a = "Pradeep"
var = pd.Series(a)
print(var,type(var))
```

```python
# creating series object by bytes type
a = bytes(100)
var = pd.Series(a)
print(var,type(var))
```

```python
# creating series object by bytearray type
a = bytearray([10,20,30])
print(a,type(a))
var = pd.Series(a)
print(var,type(var))
```

## Data Filtering and Conditional Change 

```python
# filter data whose percentage greater than 75 
df[df['percentage']>75] 
```

```python
# filter data  whose percentage between 60 and 75 
df.loc[(df['percentage']< 75)  & (df['percentage'] >=60)]
```

```python
# filtering data whose percentage less than 60 
df.loc[df["percentage"]<60]
```

```python
# now assigning grades with some rules
# Grade = A whose percentage greater than 75
# grade = B whose percentage between 60 and 75 
# grade = c whose percentage less than 60 
df["Grade"]="0"
df.loc[df['percentage']>=75,['Grade']]="A"
df.loc[(df['percentage']< 75)  & (df['percentage'] >=60),['Grade']] = "B"
df.loc[df["percentage"]<60,['Grade']]="C"
```

# GroupBy
        The Groupby mechanism in the pandas provides a way to break a dataframe into different groups or chunks based on the values of 
        single or  multiple columns

```python
df=pd.read_csv("employee.csv")
```

```python
df
```

```python
# group data on the behalf of city 
df.groupby("City")
```

```python
data = df.groupby("City")
print("Types of data :-",type(data))
```

```python
for val in data:
    print(val,type(val))
    print("-------------------------------")
```

```python
for val in data :
    print("group name :-",val[0])
    print("*"*50)
    print(val[1])
    print("*"*50)
```
